```python
def categories_athlete(row):

    if row["height_cm"] < 175 and row["weight_kg"] < 70:
        return "lightweight"

    elif row["height_cm"] < 185 or row["weight_kg"] <= 80:
        return "mediumweight"

    else:
        return "heavyweight"


bios_new["category"] = bios_new.apply(categories_athlete, axis=1)
```
